```python
from torchvision.datasets import ImageFolder
from collections import Counter

DATA_DIR = r"D:\Desktop\ml_lab\PlantVillage"

dataset = ImageFolder(DATA_DIR)

print("Number of images:", len(dataset))
print("Number of classes:", len(dataset.classes))

print("\nClass names:")
print(dataset.classes)

print("\nImages per class:")
counts = Counter(dataset.targets)

for class_index, class_name in enumerate(dataset.classes):
    print(f"{class_name}: {counts[class_index]}")
```

```python
import os

nested_path = os.path.join(DATA_DIR, "PlantVillage")

print("Nested folder exists:", os.path.isdir(nested_path))
print("\nContents of nested folder:")

if os.path.isdir(nested_path):
    print(os.listdir(nested_path))
```
